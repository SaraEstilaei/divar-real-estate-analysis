```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv('Divar.csv')
display(df.head(15))
display(df.info())
display(df.describe())
```

```python
my_columns = ['cat2_slug', 'cat3_slug', 'city_slug', 'neighborhood_slug', 'created_at_month', 'location_latitude', 'location_longitude', 'location_radius', 'property_type', 'user_type', 'title', 'description', 'regular_person_capacity', 'extra_person_capacity', 'cost_per_extra_person']
my_df = df[my_columns].copy()
print(f"Number of rows : {my_df.shape[0]:,} | Number of columns : {my_df.shape[1]}")
print('Missing values % :')
missing_percentage = my_df.isnull().mean() * 100
print(missing_percentage.round(2))
display(my_df.head())
display(my_df.describe())
```

```python
# Iran location
lat_min, lat_max = 25.06, 39.78
lon_min, lon_max = 44.03, 63.32


mask = ((my_df['location_latitude'] < lat_min) | (my_df['location_latitude'] > lat_max) |(my_df['location_longitude'] < lon_min) | (my_df['location_longitude'] > lon_max))
my_df.loc[mask, ['location_latitude', 'location_longitude']] = np.nan


neighborhood_medians = my_df.groupby(['city_slug', 'neighborhood_slug'])[['location_latitude', 'location_longitude']].transform('median')

my_df['location_latitude'] = my_df['location_latitude'].fillna(neighborhood_medians['location_latitude'])
my_df['location_longitude'] = my_df['location_longitude'].fillna(neighborhood_medians['location_longitude'])

city_medians = my_df.groupby('city_slug')[['location_latitude', 'location_longitude']].transform('median')

my_df['location_latitude'] = my_df['location_latitude'].fillna(city_medians['location_latitude'])
my_df['location_longitude'] = my_df['location_longitude'].fillna(city_medians['location_longitude'])



cols_to_drop = ['cost_per_extra_person', 'extra_person_capacity', 'regular_person_capacity', 'property_type', 'location_radius', 'neighborhood_slug']

missing_percentage = my_df.isnull().mean() * 100
display(missing_percentage.round(2))
my_df = my_df.drop(cols_to_drop, axis = 1)
display(my_df.head())
```

```python
my_df['created_at_dt'] = pd.to_datetime(my_df['created_at_month'])

min_date = my_df['created_at_dt'].min()
my_df['days_since_start'] = (my_df['created_at_dt'] - min_date).dt.days

my_df['month'] = my_df['created_at_dt'].dt.month

my_df['month_sin'] = np.sin(2 * np.pi * my_df['month'] / 12)
my_df['month_cos'] = np.cos(2 * np.pi * my_df['month'] / 12)


my_df.drop(columns=['created_at_month', 'created_at_dt'], inplace=True)


city_freq = my_df['city_slug'].value_counts()
my_df['city_freq_enc'] = my_df['city_slug'].map(city_freq)

print("cat2_slug :")
print(df['cat2_slug'].value_counts(dropna=False))
print("Number of unique categories in cat2_slug :", df['cat2_slug'].nunique())

print("cat3_slug : ")
print(df['cat3_slug'].value_counts(dropna=False))
print("Number of unique categories in cat3_slug :", df['cat3_slug'].nunique())

my_df['cat3_slug'] = my_df['cat3_slug'].fillna('unknown')
my_df['cat2_slug'] = my_df['cat2_slug'].fillna('unknown')

def extract_deal_type(cat2):
    if 'sell' in cat2:
        return 'sell'
    elif 'temporary' in cat2:
        return 'temporary_rent'
    elif 'rent' in cat2:
        return 'rent'
    else:
        return 'other'

my_df['deal_type'] = my_df['cat2_slug'].apply(extract_deal_type)

def extract_property_category(cat3):
    if 'apartment' in cat3 or 'suite' in cat3:
        return 'apartment'
    elif 'house-villa' in cat3:
        return 'villa'
    elif 'plot-old' in cat3:
        return 'plot_land'
    elif 'shop' in cat3:
        return 'shop'
    elif 'office' in cat3:
        return 'office'
    elif 'services' in cat3:
        return 'services'
    else:
        return 'other'

my_df['property_category'] = my_df['cat3_slug'].apply(extract_property_category)

cols_need_encoding = ['deal_type', 'property_category', 'user_type']

my_df_encoded = pd.get_dummies(my_df, columns=cols_need_encoding, dtype=int)
my_df_encoded.drop(columns=['cat2_slug', 'cat3_slug'], inplace=True)
print(my_df_encoded.shape)
display(my_df_encoded.head(3))
```
