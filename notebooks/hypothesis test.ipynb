```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
from scipy import stats
```

```python
df_clean = pd.read_parquet('../preprocessing_output_v4.parquet')

pd.set_option('display.max_columns', None)
display(df_clean.head())
```

```python
residential_mask = (df_clean['property_category_apartment'] == 1) | (df_clean['property_category_villa'] == 1)
df_res = df_clean[residential_mask].copy()
df_res = df_res.dropna(subset=['construction_year', 'building_size'])

lower_limit = 25
upper_limit = df_res['building_size'].quantile(0.995)

df_res_filtered = df_res[(df_res['building_size'] >= lower_limit) &  (df_res['building_size'] <= upper_limit)]
print(f"Total residential records: {len(df_res)}")
print(f"Filter range: [{lower_limit}, {upper_limit}]")
print(f"Removed extreme records: {len(df_res) - len(df_res_filtered)} ({((len(df_res) - len(df_res_filtered)) / len(df_res)) * 100}%)\n")

old_res = df_res_filtered[df_res_filtered['construction_year'] < 1396]['building_size']
new_res = df_res_filtered[df_res_filtered['construction_year'] >= 1396]['building_size']

print("=" * 50)
print("DESCRIPTIVE STATISTICS (RESIDENTIAL: APARTMENT + VILLA)")
print(f"Old Houses (<1396):")
print(f"Count (n) : {len(old_res)}")
print(f"Mean : {old_res.mean()} ")
print(f"Median : {old_res.median()} ")
print(f"Std Dev : {old_res.std(ddof=1)}")
print(f"IQR (25%-75%): [{old_res.quantile(0.25)}, {old_res.quantile(0.75)}]")

print(f"\nNew Houses (>=1396):")
print(f"Count (n) :{len(new_res)}")
print(f"Mean  : {new_res.mean()} ")
print(f"Median : {new_res.median()} ")
print(f"Std Dev : {new_res.std(ddof=1)}")
print(f"IQR (25%-75%): [{new_res.quantile(0.25)}, {new_res.quantile(0.75)}]")

t_stat, p_val = stats.ttest_ind(old_res,  new_res,  equal_var=True,  alternative='greater')
alpha = 0.05
print("HYPOTHESIS TEST RESULTS (H0: old <= new vs Ha: old > new)")
print("=" * 50)
print(f"T-statistic : {t_stat}")
print(f"P-value : {p_val}")

if p_val < alpha:
    print(f"\nResult: Reject H0 at alpha={alpha} -> Older residential units are significantly more spacious.")
else:
    print(f"\nResult: Fail to reject H0 at alpha={alpha} -> Insufficient evidence that older residential units are more spacious.")
```
