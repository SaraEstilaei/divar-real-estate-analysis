```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv('Divar.csv')
display(df.head())
```

```python
df_copy = (df[['land_size', 'building_size', 'floor', 'rooms_count', 'total_floors_count',
'unit_per_floor', 'construction_year', 'is_rebuilt', 'deed_type', 'has_business_deed']])
display(df_copy.head(30))

print(df_copy.isnull().mean()*100)
```

```python
df_copy["building_size"] = pd.to_numeric(df_copy["building_size"])
df_copy.loc[df_copy["building_size"] <= 0, "building_size"] = np.nan
df_copy["building_size"] = df_copy["building_size"].fillna(df_copy["building_size"].median())
df_copy["land_size"] = pd.to_numeric(df_copy["land_size"])
df_copy.loc[df_copy["land_size"] <= 0, "land_size"] = np.nan




```

```python
df_copy['floor'] = pd.to_numeric(df_copy['floor'], errors='coerce')
df_copy.loc[~df_copy['floor'].between(-1, 50), 'floor'] = np.nan
df_copy['floor_was_missing'] = df_copy['floor'].isna().astype(int)
df_copy['floor'] = df_copy['floor'].fillna(round(df_copy['floor'].median()))
rooms_dict = {'بدون اتاق': 0,
    'یک': 1,
    'دو': 2,
    'سه': 3,
    'چهار': 4,
    'پنج': 5,
    'شش': 6,
    'هفت': 7,
    'هشت': 8,
    'نه': 9,
    'ده': 10
}
df_copy['rooms_count'] = df_copy['rooms_count'].map(rooms_dict)
df_copy['rooms_count_was_missing'] = df_copy['rooms_count'].isna().astype(int)
df_copy['rooms_count'] = df_copy['rooms_count'].fillna(df_copy['rooms_count'].mode()[0])
display(df_copy.head())
```

```python
df_copy['total_floors_count'] = pd.to_numeric(df_copy['total_floors_count'], errors='coerce')
df_copy.loc[(df_copy['total_floors_count'] <= 0) | (df_copy['total_floors_count'] > 20), 'total_floors_count'] = np.nan
df_copy['total_floors_count_was_missing'] = df_copy['total_floors_count'].isna().astype(int)
df_copy['unit_per_floor'] = pd.to_numeric(df_copy['unit_per_floor'], errors='coerce')
df_copy.loc[(df_copy['unit_per_floor'] <= 0) | (df_copy['unit_per_floor'] > 50), 'unit_per_floor'] = np.nan
df_copy['unit_per_floor_was_missing'] = df_copy['unit_per_floor'].isna().astype(int)
year_replacements = {'قبل از ۱۳۷۰': 1365, 'قبل از 1370': 1365,}
df_copy['construction_year'] = df_copy['construction_year'].replace(year_replacements)
df_copy['construction_year'] = (df_copy['construction_year'].astype(str).str.translate(str.maketrans('۰۱۲۳۴۵۶۷۸۹', '0123456789')))
df_copy['construction_year'] = pd.to_numeric(df_copy['construction_year'], errors='coerce')
df_copy['construction_year_was_missing'] = df_copy['construction_year'].isna().astype(int)
Current_year = 1405
df_copy['building_age'] = Current_year-df_copy['construction_year']
df_copy['is_rebuilt'] = df_copy['is_rebuilt'].fillna(False).astype(bool).astype(int)
df_copy['deed_type'] = df_copy['deed_type'].fillna('unknown').astype(str).str.strip().str.lower()
df_copy['has_business_deed_was_missing'] = df_copy['has_business_deed'].isna().astype(int)
df_copy['has_business_deed_encoded'] = df_copy['has_business_deed'].map({True: 1, False: 0}).fillna(-1).astype(int)
display(df_copy.head(30))
```
