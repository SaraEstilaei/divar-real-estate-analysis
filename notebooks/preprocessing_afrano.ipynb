```python
import pandas as pd
df = pd.read_parquet(r"C:\Users\Afrano\Documents\Quera\ML project\preprocessing_output_v4.parquet")
display(df.head())
```

```python
df_copy = df.copy()

selected_columns = [
    'land_size', 'building_size', 'floor', 'rooms_count', 'total_floors_count',
    'unit_per_floor', 'construction_year', 'is_rebuilt', 'deed_type', 'has_business_deed'
]

display(df_copy[selected_columns].head(10))
print(df_copy[selected_columns].isnull().mean() * 100)
```

```python
# building_size
df_copy['building_size'] = pd.to_numeric(df_copy['building_size'], errors='coerce')
df_copy.loc[df_copy['building_size'] <= 0, 'building_size'] = np.nan
df_copy['building_size_was_missing'] = df_copy['building_size'].isna().astype(int)
df_copy['building_size'] = df_copy['building_size'].fillna(df_copy['building_size'].median())

# land_size
df_copy['land_size'] = pd.to_numeric(df_copy['land_size'], errors='coerce')
df_copy.loc[df_copy['land_size'] <= 0, 'land_size'] = np.nan
df_copy['land_size_was_missing'] = df_copy['land_size'].isna().astype(int)
df_copy['land_size'] = df_copy['land_size'].fillna(df_copy['land_size'].median())

display(df_copy[['building_size', 'building_size_was_missing', 'land_size', 'land_size_was_missing']].head())




```

```python
# floor
df_copy['floor'] = pd.to_numeric(df_copy['floor'], errors='coerce')
df_copy.loc[(df_copy['floor'] < -2) | (df_copy['floor'] > 100), 'floor'] = np.nan
df_copy['floor_was_missing'] = df_copy['floor'].isna().astype(int)
df_copy['floor'] = df_copy['floor'].fillna(round(df_copy['floor'].median()))

# rooms_count
rooms_dict = {
    'بدون اتاق': 0, 'یک': 1, 'دو': 2, 'سه': 3, 'چهار': 4,
    'پنج': 5, 'شش': 6, 'هفت': 7, 'هشت': 8, 'نه': 9, 'ده': 10
}
df_copy['rooms_count'] = df_copy['rooms_count'].replace(rooms_dict)
df_copy['rooms_count'] = pd.to_numeric(df_copy['rooms_count'], errors='coerce')
df_copy.loc[(df_copy['rooms_count'] < 0) | (df_copy['rooms_count'] > 20), 'rooms_count'] = np.nan
df_copy['rooms_count_was_missing'] = df_copy['rooms_count'].isna().astype(int)
df_copy['rooms_count'] = df_copy['rooms_count'].fillna(df_copy['rooms_count'].mode()[0])

display(df_copy[['floor', 'floor_was_missing', 'rooms_count', 'rooms_count_was_missing']].head())
```

```python
# total_floors_count
df_copy['total_floors_count'] = pd.to_numeric(df_copy['total_floors_count'], errors='coerce')
df_copy.loc[(df_copy['total_floors_count'] <= 0) | (df_copy['total_floors_count'] > 120), 'total_floors_count'] = np.nan
df_copy['total_floors_count_was_missing'] = df_copy['total_floors_count'].isna().astype(int)
df_copy['total_floors_count'] = df_copy['total_floors_count'].fillna(round(df_copy['total_floors_count'].median()))

# unit_per_floor
df_copy['unit_per_floor'] = pd.to_numeric(df_copy['unit_per_floor'], errors='coerce')
df_copy.loc[(df_copy['unit_per_floor'] <= 0) | (df_copy['unit_per_floor'] > 100), 'unit_per_floor'] = np.nan
df_copy['unit_per_floor_was_missing'] = df_copy['unit_per_floor'].isna().astype(int)
df_copy['unit_per_floor'] = df_copy['unit_per_floor'].fillna(round(df_copy['unit_per_floor'].median()))

# construction_year & building_age
year_replacements = {'قبل از ۱۳۷۰': 1365, 'قبل از 1370': 1365}
df_copy['construction_year'] = df_copy['construction_year'].replace(year_replacements)
df_copy['construction_year'] = (df_copy['construction_year'].astype(str).str.translate(str.maketrans('۰۱۲۳۴۵۶۷۸۹', '0123456789')))
df_copy['construction_year'] = pd.to_numeric(df_copy['construction_year'], errors='coerce')
df_copy.loc[(df_copy['construction_year'] < 1300) | (df_copy['construction_year'] > 1404), 'construction_year'] = np.nan
df_copy['construction_year_was_missing'] = df_copy['construction_year'].isna().astype(int)
df_copy['construction_year'] = df_copy['construction_year'].fillna(round(df_copy['construction_year'].median()))

current_year = 1404
df_copy['building_age'] = (current_year - df_copy['construction_year']).clip(lower=0)

# is_rebuilt (keep missing as separate category: 1=True, 0=False, -1=missing)
df_copy['is_rebuilt_was_missing'] = df_copy['is_rebuilt'].isna().astype(int)
df_copy['is_rebuilt'] = df_copy['is_rebuilt'].map({True: 1, False: 0}).fillna(-1).astype(int)

# deed_type
df_copy['deed_type_was_missing'] = df_copy['deed_type'].isna().astype(int)
df_copy['deed_type'] = df_copy['deed_type'].fillna('unknown').astype(str).str.strip().str.lower()

# has_business_deed
df_copy['has_business_deed_was_missing'] = df_copy['has_business_deed'].isna().astype(int)
df_copy['has_business_deed'] = df_copy['has_business_deed'].map({True: 1, False: 0}).fillna(-1).astype(int)

display(df_copy[['total_floors_count', 'unit_per_floor', 'building_age', 'is_rebuilt', 'has_business_deed']].head())
```

```python
output_path = r"C:\Users\Afrano\Documents\Quera\ML project\preprocessing_output_v4.parquet"
df_copy.to_parquet(output_path, index=False)
print("Saved successfully to:", output_path)
print("Total rows and columns:", df_copy.shape)
```
